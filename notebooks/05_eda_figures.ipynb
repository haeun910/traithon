```python
# 제목 본문 유사도 분포 히스토그램
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from matplotlib import font_manager

# -----------------------------
# 한글 폰트 설정
# -----------------------------
plt.rcParams["axes.unicode_minus"] = False
available_fonts = {f.name for f in font_manager.fontManager.ttflist}
for font in ["Malgun Gothic", "AppleGothic", "NanumGothic"]:
    if font in available_fonts:
        plt.rcParams["font.family"] = font
        break

# -----------------------------
# 데이터: AI Hub 낚시성 기사 라벨링 ZIP(JSON) → newsTitle, newsContent
# -----------------------------
import json, zipfile
ZIP_PATH = "../data/raw/clickbait_labels/VL_Part1_Clickbait_Auto_EC.zip"
rows = []
with zipfile.ZipFile(ZIP_PATH) as zf:
    for name in zf.namelist():
        if not name.endswith(".json"):
            continue
        obj = json.loads(zf.read(name).decode("utf-8", errors="ignore"))
        src = obj.get("sourceDataInfo", obj)
        rows.append({"newsTitle": src.get("newsTitle"), "newsContent": src.get("newsContent")})
df = pd.DataFrame(rows)
df["newsTitle"] = df["newsTitle"].fillna("")
df["newsContent"] = df["newsContent"].fillna("")

# -----------------------------
# 제목-본문 유사도 계산
# -----------------------------
vectorizer = TfidfVectorizer(max_features=5000)
title_vec = vectorizer.fit_transform(df["newsTitle"])
content_vec = vectorizer.transform(df["newsContent"])

# 행끼리만 비교 (N×N 전체 행렬을 만들지 않음). TF-IDF 벡터는 L2 정규화되어 있어 내적 = 코사인 유사도
df["title_content_sim"] = title_vec.multiply(content_vec).sum(axis=1).A1

# -----------------------------
# 시각화
# -----------------------------
plt.figure(figsize=(8, 5))
plt.hist(df["title_content_sim"], bins=30)
plt.title("제목–본문 유사도 분포", fontsize=15)
plt.xlabel("Cosine Similarity (Title vs Content)")
plt.ylabel("기사 수")
plt.tight_layout()
plt.savefig("../figures/impact2_title_content_similarity.png", dpi=300)
plt.show()
```
